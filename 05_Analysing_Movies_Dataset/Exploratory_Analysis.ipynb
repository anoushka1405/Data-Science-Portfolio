## Most Successful Actors

```python
df.cast[0]
```

```python
df.cast.str.split("|", expand = True).stack()
```

```python
df.cast.str.split("|", expand=True).stack().dropna()
```

```python
act = (df.cast
       .str.split("|", expand=True)
       .stack()
       .dropna()
       .reset_index(level=1, drop=True)
       .to_frame()
      )

act.columns = ["actor"]
```

```python
act
```

```python
act = act.merge(
    df[["revenue_musd", "vote_average"]],
    left_index=True,
    right_index=True
)
```

```python
act
```

```python
act.groupby("actor").agg(
    total_revenue = ("revenue_musd", "sum"),
    mean_rating   = ("vote_average", "mean"),
    total_movies  = ("actor",        "count")
).sort_values("total_revenue", ascending=False).head(20)
```

Stan Lee and Samuel L. Jackson top the list by total revenue, largely due to their involvement in the Marvel Cinematic Universe, one of the highest grossing franchises in history. However, total revenue alone is a misleading metric for actor "success" since it rewards quantity of appearances over quality. An actor in 100 average films will outscore one in 10 masterpieces.

```python
act.groupby("actor").agg(
    total_revenue = ("revenue_musd", "sum"),
    mean_rating   = ("vote_average", "mean"),
    total_movies  = ("actor",        "count")
).query("total_movies >= 30").sort_values("mean_rating", ascending=False).head(20)
```

Requiring a minimum of 30 appearances filters out actors who had a lucky run of good films and surfaces those with long, sustained careers of quality work. Toshirō Mifune leads : a Japanese cinema icon who starred in 16 Kurosawa films. Notably, Martin Scorsese and Werner Herzog appear despite being primarily directors, highlighting a limitation of the dataset where cast and crew roles are not always distinguished.
